```python
import os
from dotenv import load_dotenv
load_dotenv

groq_api_key = os.getenv("GROQ_API_KEY")
```

```python
from langchain_groq import ChatGroq
model = ChatGroq(model="openai/gpt-oss-20b", groq_api_key=groq_api_key)
model
```

```python
!pip install langchain_core
```

```python
### Manually install langchain_core from source
from langchain_core.messages import SystemMessage, HumanMessage

messages = [
    SystemMessage(content="Translate the following English text to Telugu."),
    HumanMessage(content="I love programming.")
]


response = model.invoke(messages)

# print(response.content)
```

```python
from langchain_core.output_parsers import StrOutputParser

parser = StrOutputParser()
parser.invoke(response)
```

```python
chain = model | parser
chain.invoke(messages)
```

```python
### By using the ChatPromptTemplate, you can create a prompt template that can be used to generate prompts for the model. The template can include placeholders for user input, which can be filled in at runtime.
from langchain_core.prompts import ChatPromptTemplate

generic_template = "Translate the following into {language}:"

prompt = ChatPromptTemplate.from_messages(
    [("system", generic_template),
    ("user", "{text}")]
    )
```

```python
result =prompt.invoke({"language": "Telugu", "text": "I love programming."})
```

```python
result
```

```python
chain = prompt | model | parser
chain.invoke({"language": "Telugu", "text": "I love programming."})
```
